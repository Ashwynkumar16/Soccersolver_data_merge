```python
import pandas as pd

def prune_datasets(df_list):
    """
    Strips noise, text, and granular positional ratings from a list of dataframes.
    """
    # Phase 1 Drop Targets
    url_image_cols = [
        'player_url', 'player_face_url', 'club_logo_url', 
        'club_flag_url', 'nation_logo_url', 'nation_flag_url', 'image'
    ]
    
    text_cols = ['description']
    
    positional_ratings = [
        'ls', 'st', 'rs', 'lw', 'lf', 'cf', 'rf', 'rw', 'lam', 'cam', 'ram', 
        'lm', 'lcm', 'cm', 'rcm', 'rm', 'lwb', 'ldm', 'cdm', 'rdm', 'rwb', 
        'lb', 'lcb', 'cb', 'rcb', 'rb', 'gk'
    ]
    
    # Combine all targets into a single master drop list
    cols_to_drop = url_image_cols + text_cols + positional_ratings
    
    # Execute drop across all dataframes in-place
    for i in range(len(df_list)):
        df_list[i].drop(columns=cols_to_drop, errors='ignore', inplace=True)
        print(f"Dataset {i+1} reduced to {len(df_list[i].columns)} columns.")
        
    return df_list

df_list = prune_datasets(df_list)

for df in df_list:
    print(df.shape)
    print(df.columns.tolist())
    print('-'*50)
```

```python
import pandas as pd

def standardize_schema(df_list):
    """
    Unifies dataset columns to the concise 76-column schema naming convention.
    """
    # 1. Explicit mapping for core identifiers, financials, and team data
    master_mapping = {
        'sofifa_id': 'player_id',
        'short_name': 'name',
        'long_name': 'full_name',
        'player_positions': 'positions',
        'overall': 'overall_rating',
        'value_eur': 'value',
        'wage_eur': 'wage',
        'release_clause_eur': 'release_clause',
        'club_team_id': 'club_id',
        'league_name': 'club_league_name',
        'club_jersey_number': 'club_kit_number',
        'club_joined_date': 'club_joined',
        'club_contract_valid_until_year': 'club_contract_valid_until',
        'nationality_id': 'country_id',
        'nationality_name': 'country_name',
        'nation_jersey_number': 'country_kit_number',
        'player_tags': 'specialities',
        'player_traits': 'play_styles',
        # Edge cases explicitly defined
        'defending_marking_awareness': 'defensive_awareness',
        'goalkeeping_diving': 'gk_diving',
        'goalkeeping_handling': 'gk_handling',
        'goalkeeping_kicking': 'gk_kicking',
        'goalkeeping_positioning': 'gk_positioning',
        'goalkeeping_reflexes': 'gk_reflexes',
        'goalkeeping_speed': 'gk_speed'
    }

    # 2. Prefixes to dynamically strip from granular attributes
    prefixes_to_strip = [
        'attacking_', 'skill_', 'movement_', 
        'power_', 'mentality_', 'defending_'
    ]

    for i, df in enumerate(df_list):
        # NEW FIX: Isolate the macro dribbling stat before stripping prefixes
        if 'skill_dribbling' in df.columns and 'dribbling' in df.columns:
            df.rename(columns={'dribbling': 'macro_dribbling'}, inplace=True)

        # Apply the explicit master dictionary
        df.rename(columns=master_mapping, inplace=True)
        
        # Dynamically strip the prefixes from the remaining 110-col schemas
        new_columns = {}
        for col in df.columns:
            for prefix in prefixes_to_strip:
                if col.startswith(prefix):
                    new_columns[col] = col.replace(prefix, '')
                    break # Stop checking prefixes once a match is found
        
        df.rename(columns=new_columns, inplace=True)
        print(f"Dataset {i+1} schema standardized. Current column count: {len(df.columns)}")

    return df_list


df_list = standardize_schema(df_list)
def df_check():
    for df in df_list:
        print(df.shape)
        print(df.columns.tolist())
        print('-'*50)

df_check()
```

```python
def build_master_dataset(df_list):
    """
    Executes the inner join stack, removes duplicates, and generates 
    a diagnostic report of dropped columns per season.
    """
    season_labels = [
        '2025-2026', '2024-2025', '2023-2024', 
        '2022-2023', '2021-2022', '2020-2021', '2019-2020'
    ]
    
    # Store the original column sets before the inner join destroys them
    original_cols = [set(df.columns) for df in df_list]
    
    # Step 2: Inner Join Concatenation
    print("Executing Outer Join...")
    master_df = pd.concat(df_list, join='outer', ignore_index=True)
    master_col_set = set(master_df.columns)
    
    # Generate Drop Report via Set Difference
    print("\n--- Column Drop Diagnostic Report ---")
    for i, orig_set in enumerate(original_cols):
        dropped_cols = orig_set - master_col_set
        if dropped_cols:
            print(f"Season {season_labels[i]} dropped {len(dropped_cols)} columns:")
            # Sorting for readability in the terminal
            print(f" -> {sorted(list(dropped_cols))}\n")
        else:
            print(f"Season {season_labels[i]} dropped 0 columns.\n")
            
    # Step 3: Anomaly Resolution
    rows_before = len(master_df)
    master_df.drop_duplicates(subset=['player_id', 'season'], keep='first', inplace=True)
    rows_after = len(master_df)
    
    # Step 4: Index Reset
    master_df.reset_index(drop=True, inplace=True)
    
    # Final Output Summary
    print("--- Final Pipeline Stats ---")
    print(f"Rows dropped during deduplication (22/23 anomaly): {rows_before - rows_after}")
    print(f"Final Master Dataset Shape: {master_df.shape}")
    print(f"Uniform Features Available: {len(master_df.columns)}")
    
    return master_df
```

```python
# Execution:
master_df = build_master_dataset(df_list)
```

```python
def normalize_string_series(series: pd.Series) -> pd.Series:
    """Vectorized string cleaning: strips diacritics, converts to lowercase,removes punctuation/trailing white spaces, and handles nulls safely.
    """
    # Fill missing values and cast to string
    clean_series = series.fillna("").astype(str)

    # Decompose unicode characters (e.g., é -> e + ´)
    clean_series = clean_series.apply(
        lambda x: unicodedata.normalize("NFD", x)
    )

    # Encode to ASCII (dropping the dangling accents), then decode back
    clean_series = clean_series.apply(
        lambda x: x.encode("ascii", "ignore").decode("utf-8")
    )

    # Lowercase, clean punctuation artifacts, and strip white spaces
    clean_series = (
        clean_series.str.lower()
        .str.replace(r"[^\w\s]", "", regex=True)
        .str.strip()
    )

    return clean_series
```

```python
print("LAUNCHING PHASE 1: PRE-FLIGHT STANDARDIZATION")
# print("=" * 60)

if not os.path.exists(SSWY_INPUT_PATH) or not os.path.exists(SOFIFA_INPUT_PATH):
    print("ERROR: Input file paths are incorrect. Please update Cell 1.")
else:
    # 1. Loading
    print("Loading datasets into memory...")
    df_sswy = pd.read_csv(SSWY_INPUT_PATH)
    df_sofifa = pd.read_csv(SOFIFA_INPUT_PATH)

    # 2. Namespace Protection
    print("Applying 'sofifa_' prefix to columns...")
    df_sofifa = df_sofifa.add_prefix("sofifa_")
    print("Enforcing ID string types to prevent rolling dictionary failures...")
    
    # #UPDATE: Force SSWY IDs to string and strip float artifacts
    df_sswy['soccersolver_id'] = df_sswy['soccersolver_id'].astype(str).str.replace(r'\.0$', '', regex=True)
    # #UPDATE: Force Sofifa IDs to string and strip float artifacts
    df_sofifa['sofifa_player_id'] = df_sofifa['sofifa_player_id'].astype(str).str.replace(r'\.0$', '', regex=True)

    # 3. Deduplication (Sofifa Version Control)
    print("Isolating definitive seasonal roster updates...")
    df_sofifa["sofifa_fifa_update_date"] = pd.to_datetime(df_sofifa["sofifa_fifa_update_date"])
    df_sofifa = df_sofifa.sort_values(by="sofifa_fifa_update_date", ascending=True)
    
    initial_sofifa_count = len(df_sofifa)
    df_sofifa = df_sofifa.drop_duplicates(subset=["sofifa_player_id", "sofifa_season"], keep="last")
    print(f" -> Removed {initial_sofifa_count - len(df_sofifa)} transient roster updates.")

    # 4. Text Normalization & Mapping
    print("Normalizing names and standardizing country aliases...")
    df_sswy["normalized_name"] = normalize_string_series(df_sswy["name"])
    df_sofifa["normalized_name"] = normalize_string_series(df_sofifa["sofifa_full_name"])

    # Extract, lowercase, map, and clean nationalities
    raw_sswy_nat = df_sswy["nationality"].fillna("").astype(str).str.lower().str.strip()
    raw_sofifa_nat = df_sofifa["sofifa_country_name"].fillna("").astype(str).str.lower().str.strip()

    df_sswy["normalized_nationality"] = normalize_string_series(
        raw_sswy_nat.map(lambda x: COUNTRY_MAP.get(x, x))
    )
    df_sofifa["normalized_nationality"] = normalize_string_series(
        raw_sofifa_nat.map(lambda x: COUNTRY_MAP.get(x, x))
    )

    # 5. Temporal Alignment & Block Extraction
    print("Aligning temporal data and extracting evaluation blocks...")
    df_sswy["birth_date"] = pd.to_datetime(df_sswy["birth_date"], errors="coerce")
    df_sofifa["sofifa_dob"] = pd.to_datetime(df_sofifa["sofifa_dob"], errors="coerce")

    df_sswy["clean_dob"] = df_sswy["birth_date"].dt.strftime("%Y-%m-%d")
    df_sofifa["clean_dob"] = df_sofifa["sofifa_dob"].dt.strftime("%Y-%m-%d")

    df_sswy["Birth_Year"] = df_sswy["birth_date"].dt.year.fillna(0).astype(int)
    df_sofifa["Birth_Year"] = df_sofifa["sofifa_dob"].dt.year.fillna(0).astype(int)

    df_sswy["name_first_letter"] = df_sswy["normalized_name"].str[0].fillna("")
    df_sofifa["name_first_letter"] = df_sofifa["normalized_name"].str[0].fillna("")

    # print("-" * 60)
    print("PHASE 1 COMPLETE")
    print(f"Final Counts -> SSWY: {len(df_sswy)} rows | Sofifa: {len(df_sofifa)} rows")
    print("DataFrames are currently held in memory. Ready for Phase 2.")
    # print("=" * 60)
```

```python
from thefuzz import fuzz
import pandas as pd
import numpy as np

def calculate_match_score(row) -> pd.Series:
    """
    Executes the 100-point scoring matrix. Thresholds: >= 80 (Matched), 72-79.9 (Partial), < 72 (Discard)
    """
    score = 0.0
    
    # 1. Name Match (60 Points) - using token_set_ratio for subset neutralization
    name_sswy = str(row.get('normalized_name_sswy', ''))
    name_sof = str(row.get('normalized_name_sof', ''))
    
    if name_sswy and name_sof:
        score += fuzz.token_set_ratio(name_sswy, name_sof) * 0.60

    # 2. Nationality Match (25 Points)
    nat_sswy = str(row.get('normalized_nationality_sswy', 'nan'))
    nat_sof = str(row.get('normalized_nationality_sof', 'nan'))
    if nat_sswy != 'nan' and nat_sof != 'nan' and nat_sswy == nat_sof:
        score += 25.0

    # 3. Height Match (7.5 Points)
    h_sswy = row.get('height_cm')
    h_sof = row.get('sofifa_height_cm')
    if pd.notna(h_sswy) and pd.notna(h_sof):
        if abs(float(h_sswy) - float(h_sof)) <= 3.0:
            score += 7.5

    # 4. Preferred Foot (7.5 Points)
    foot_sswy = str(row.get('foot', 'nan')).lower()
    foot_sof = str(row.get('sofifa_preferred_foot', 'nan')).lower()
    if foot_sswy != 'nan' and foot_sof != 'nan' and foot_sswy == foot_sof:
        score += 7.5

    # Determine Routing Status
    if score >= 80.0:
        status = 'MATCHED'
    elif score >= 72.0:
        status = 'PARTIAL'
    else:
        status = 'DISCARD'
        
    return pd.Series([score, status], index=['match_score', 'match_status'])
```

```python
print("LAUNCHING PHASE 2: SNOWBALL IDENTITY RESOLUTION ENGINE")
# print("=" * 80)

final_matched_dfs = []
final_partial_dfs = []
final_unmatched_sswy_dfs = []
final_unmatched_sofifa_dfs = []

# Initialize the empty Rolling Dictionary
rolling_id_dictionary = pd.DataFrame(columns=['soccersolver_id', 'sofifa_player_id'])

# Sort seasons chronologically to ensure the snowball effect works
seasons = sorted(df_sswy['season'].dropna().unique())
total_seasons = len(seasons)

for idx, season in enumerate(seasons, 1):
    print(f"\n[{idx}/{total_seasons}] LIVE TRACKER: Season {season}")
    print("-" * 80)
    
    curr_sswy = df_sswy[df_sswy['season'] == season].copy()
    curr_sof = df_sofifa[df_sofifa['sofifa_season'] == season].copy()
    
    initial_sswy = len(curr_sswy)
    initial_sof = len(curr_sof)
    
    season_matched = 0
    season_partial = 0
    
    # -------------------------------------------------------------------------
    # PASS 0: The VIP Bypass (Rolling Dictionary)
    # -------------------------------------------------------------------------
    p0_matched_count = 0
    if not rolling_id_dictionary.empty:
        valid_map = rolling_id_dictionary[
            rolling_id_dictionary['soccersolver_id'].isin(curr_sswy['soccersolver_id']) & 
            rolling_id_dictionary['sofifa_player_id'].isin(curr_sof['sofifa_player_id'])
        ]
        
        if not valid_map.empty:
            p0_sswy = curr_sswy[curr_sswy['soccersolver_id'].isin(valid_map['soccersolver_id'])].copy()
            p0_sof = curr_sof[curr_sof['sofifa_player_id'].isin(valid_map['sofifa_player_id'])].copy()
            
            p0_matched = p0_sswy.merge(valid_map, on='soccersolver_id').merge(p0_sof, on='sofifa_player_id', suffixes=('_sswy', '_sof'))
            p0_matched['match_score'] = 100.0
            p0_matched['match_status'] = 'MATCHED'
            
            p0_matched = p0_matched.drop_duplicates(subset=['sofifa_player_id'], keep='first')
            p0_matched = p0_matched.drop_duplicates(subset=['soccersolver_id'], keep='first')
            
            p0_matched_count = len(p0_matched)
            season_matched += p0_matched_count
            final_matched_dfs.append(p0_matched)
            
            # Remove bypassed players from the working pool
            curr_sswy = curr_sswy[~curr_sswy['soccersolver_id'].isin(valid_map['soccersolver_id'])]
            curr_sof = curr_sof[~curr_sof['sofifa_player_id'].isin(valid_map['sofifa_player_id'])]
            
    print(f"  -> Pass 0 (ID Bypass):      {p0_matched_count} Matched (Current Dictionary Size: {len(rolling_id_dictionary)})")
    print(f"     [Rookies/Leftovers pushed to Heuristics: {len(curr_sof)} Sofifa Players]")

    # -------------------------------------------------------------------------
    # PASS 1: The Exact Anchor (Date of Birth)
    # -------------------------------------------------------------------------
    p1_merge = curr_sswy.merge(curr_sof, on='clean_dob', how='inner', suffixes=('_sswy', '_sof'))
    
    p1_matched_count = p1_partial_count = 0
    p1_matched = pd.DataFrame()
    if not p1_merge.empty:
        p1_merge[['match_score', 'match_status']] = p1_merge.apply(calculate_match_score, axis=1)
        
        p1_merge = p1_merge.sort_values('match_score', ascending=False)
        p1_merge = p1_merge.drop_duplicates(subset=['sofifa_player_id'], keep='first')
        p1_merge = p1_merge.drop_duplicates(subset=['soccersolver_id'], keep='first')
        
        p1_matched = p1_merge[p1_merge['match_status'] == 'MATCHED']
        p1_partial = p1_merge[p1_merge['match_status'] == 'PARTIAL']
        
        p1_matched_count = len(p1_matched)
        p1_partial_count = len(p1_partial)
        season_matched += p1_matched_count
        season_partial += p1_partial_count
        
        final_matched_dfs.append(p1_matched)
        final_partial_dfs.append(p1_partial)
        
        routed_sswy_ids = pd.concat([p1_matched['soccersolver_id'], p1_partial['soccersolver_id']])
        routed_sof_ids = pd.concat([p1_matched['sofifa_player_id'], p1_partial['sofifa_player_id']])
        
        curr_sswy = curr_sswy[~curr_sswy['soccersolver_id'].isin(routed_sswy_ids)]
        curr_sof = curr_sof[~curr_sof['sofifa_player_id'].isin(routed_sof_ids)]
        
    print(f"  -> Pass 1 (Exact DOB):      {p1_matched_count} Matched | {p1_partial_count} Partial")

    # -------------------------------------------------------------------------
    # PASS 2: The Nationality Net (Birth Year + Nationality)
    # -------------------------------------------------------------------------
    p2_merge = curr_sswy.merge(curr_sof, on=['Birth_Year', 'normalized_nationality'], how='inner', suffixes=('_sswy', '_sof'))
    
    p2_matched_count = p2_partial_count = 0
    p2_matched = pd.DataFrame()
    if not p2_merge.empty:
        p2_merge['normalized_nationality_sswy'] = p2_merge['normalized_nationality']
        p2_merge['normalized_nationality_sof'] = p2_merge['normalized_nationality']
        
        p2_merge[['match_score', 'match_status']] = p2_merge.apply(calculate_match_score, axis=1)
        
        p2_merge = p2_merge.sort_values('match_score', ascending=False)
        p2_merge = p2_merge.drop_duplicates(subset=['sofifa_player_id'], keep='first')
        p2_merge = p2_merge.drop_duplicates(subset=['soccersolver_id'], keep='first')
        
        p2_matched = p2_merge[p2_merge['match_status'] == 'MATCHED']
        p2_partial = p2_merge[p2_merge['match_status'] == 'PARTIAL']
        
        p2_matched_count = len(p2_matched)
        p2_partial_count = len(p2_partial)
        season_matched += p2_matched_count
        season_partial += p2_partial_count
        
        final_matched_dfs.append(p2_matched)
        final_partial_dfs.append(p2_partial)
        
        routed_sswy_ids = pd.concat([p2_matched['soccersolver_id'], p2_partial['soccersolver_id']])
        routed_sof_ids = pd.concat([p2_matched['sofifa_player_id'], p2_partial['sofifa_player_id']])
        
        curr_sswy = curr_sswy[~curr_sswy['soccersolver_id'].isin(routed_sswy_ids)]
        curr_sof = curr_sof[~curr_sof['sofifa_player_id'].isin(routed_sof_ids)]
        
    print(f"  -> Pass 2 (Nationality):    {p2_matched_count} Matched | {p2_partial_count} Partial")

    # -------------------------------------------------------------------------
    # PASS 3: The Dual-National Safety Net (Birth Year + First Letter)
    # -------------------------------------------------------------------------
    p3_merge = curr_sswy.merge(curr_sof, on=['Birth_Year', 'name_first_letter'], how='inner', suffixes=('_sswy', '_sof'))
    
    p3_matched_count = p3_partial_count = 0
    if not p3_merge.empty:
        p3_merge[['match_score', 'match_status']] = p3_merge.apply(calculate_match_score, axis=1)
        
        p3_merge = p3_merge.sort_values('match_score', ascending=False)
        p3_merge = p3_merge.drop_duplicates(subset=['sofifa_player_id'], keep='first')
        p3_merge = p3_merge.drop_duplicates(subset=['soccersolver_id'], keep='first')
        
        # GUARDRAIL: Ban Pass 3 from Matched (Forces manual review)
        p3_merge.loc[p3_merge['match_status'] == 'MATCHED', 'match_status'] = 'PARTIAL'
        p3_partial = p3_merge[p3_merge['match_status'] == 'PARTIAL']
        
        p3_partial_count = len(p3_partial)
        season_partial += p3_partial_count
        final_partial_dfs.append(p3_partial)
        
        routed_sswy_ids = p3_partial['soccersolver_id']
        routed_sof_ids = p3_partial['sofifa_player_id']
        
        curr_sswy = curr_sswy[~curr_sswy['soccersolver_id'].isin(routed_sswy_ids)]
        curr_sof = curr_sof[~curr_sof['sofifa_player_id'].isin(routed_sof_ids)]
        
    print(f"  -> Pass 3 (Safety Net):     0 Matched (Safeguard) | {p3_partial_count} Partial")
    
    # -------------------------------------------------------------------------
    # DICTIONARY UPDATE
    # -------------------------------------------------------------------------
    # Append strictly the PERFECT MATCHES from Pass 1 and Pass 2 to the dictionary
    new_perfect_ids = []
    if not p1_matched.empty:
        new_perfect_ids.append(p1_matched[['soccersolver_id', 'sofifa_player_id']])
    if not p2_matched.empty:
        new_perfect_ids.append(p2_matched[['soccersolver_id', 'sofifa_player_id']])
        
    if new_perfect_ids:
        new_pairs_df = pd.concat(new_perfect_ids, ignore_index=True)
        rolling_id_dictionary = pd.concat([rolling_id_dictionary, new_pairs_df], ignore_index=True)
        rolling_id_dictionary = rolling_id_dictionary.drop_duplicates()
        print(f"  [+] Added {len(new_pairs_df)} newly discovered pairs to the Rolling Dictionary.")

    # UNMATCHED ISOLATION
    final_unmatched_sswy_dfs.append(curr_sswy)
    final_unmatched_sofifa_dfs.append(curr_sof)

print("\n" + "=" * 80)
print("SNOWBALL LOOP COMPLETE. EXECUTING RETROACTIVE LOOP-BACK...")
print("=" * 80)

# Concatenate all lists
df_master_matched = pd.concat(final_matched_dfs, ignore_index=True) if final_matched_dfs else pd.DataFrame()
df_master_partial = pd.concat(final_partial_dfs, ignore_index=True) if final_partial_dfs else pd.DataFrame()
df_unmatched_sswy = pd.concat(final_unmatched_sswy_dfs, ignore_index=True) if final_unmatched_sswy_dfs else pd.DataFrame()
df_unmatched_sofifa = pd.concat(final_unmatched_sofifa_dfs, ignore_index=True) if final_unmatched_sofifa_dfs else pd.DataFrame()

initial_unmatched_sswy = len(df_unmatched_sswy)

# Retroactive Rescue: Join Unmatched Data against the final matured dictionary
if not df_unmatched_sswy.empty and not df_unmatched_sofifa.empty and not rolling_id_dictionary.empty:
    
    # We must ensure we rescue records from the same season
    rescued_records = df_unmatched_sswy.merge(
        rolling_id_dictionary, on='soccersolver_id', how='inner'
    ).merge(
        df_unmatched_sofifa, left_on=['sofifa_player_id', 'season'], right_on=['sofifa_player_id', 'sofifa_season'], suffixes=('_sswy', '_sof')
    )
    
    rescued_records['match_score'] = 100.0
    rescued_records['match_status'] = 'MATCHED (RESCUED)'
    
    rescued_count = len(rescued_records)
    
    if rescued_count > 0:
        # Append rescued records to master matched
        df_master_matched = pd.concat([df_master_matched, rescued_records], ignore_index=True)
        
        # Drop rescued records from unmatched pools
        df_unmatched_sswy = df_unmatched_sswy[~df_unmatched_sswy['soccersolver_id'].isin(rescued_records['soccersolver_id'])]
        df_unmatched_sofifa = df_unmatched_sofifa[~df_unmatched_sofifa['sofifa_player_id'].isin(rescued_records['sofifa_player_id'])]
        
        print(f"Retroactive Rescue Successful: Salvaged {rescued_count} orphaned records from early seasons.")
    else:
        print("Retroactive Rescue: 0 orphans found matching the final dictionary.")

print(f"\nFinal Global Statistics:")
print(f"  - Total Matched: {len(df_master_matched)}")
print(f"  - Total Partial (Review): {len(df_master_partial)}")
print(f"  - Total Unmatched SSWY: {len(df_unmatched_sswy)}")
print(f"  - Total Unmatched Sofifa: {len(df_unmatched_sofifa)}")
```

```python
import pandas as pd
import numpy as np
from thefuzz import fuzz

# ==============================================================================
# 1. PRE-PROCESSING: ENSURE STRINGS & FIX FLOAT IDS
# ==============================================================================
# Apply your ID guardrail to the specific primary keys
df_unmatched_sswy['soccersolver_id'] = df_unmatched_sswy['soccersolver_id'].astype(str).str.replace(r'\.0$', '', regex=True)
df_unmatched_sofifa['sofifa_player_id'] = df_unmatched_sofifa['sofifa_player_id'].astype(str).str.replace(r'\.0$', '', regex=True)

# Sort seasons to guarantee chronological learning (using SSWY's season column)
seasons = sorted(df_unmatched_sswy['season'].dropna().unique())
total_seasons = len(seasons)

# ==============================================================================
# GUARDRAIL: SANITIZE THE ROLLING DICTIONARY
# ==============================================================================
# If rolling_id_dictionary is a DataFrame, convert it to a dict
if isinstance(rolling_id_dictionary, pd.DataFrame):
    # Swap 'soccersolver_id' and 'sofifa_player_id' with your actual column names if they differ in the DF
    rolling_id_dictionary = dict(zip(
        rolling_id_dictionary['soccersolver_id'].astype(str), 
        rolling_id_dictionary['sofifa_player_id'].astype(str)
    ))
elif isinstance(rolling_id_dictionary, dict):
    # If it is a dict, force all keys and values to be flat 1D strings to strip any hidden Pandas Series/Arrays
    rolling_id_dictionary = {str(k): str(v) for k, v in rolling_id_dictionary.items()}
else:
    raise TypeError(f"rolling_id_dictionary is an unexpected type: {type(rolling_id_dictionary)}")

# ==============================================================================
# 2. THE CHRONOLOGICAL SNOWBALL ENGINE
# ==============================================================================
for idx, season in enumerate(seasons, start=1):
    
    # Tracker Variables setup
    dict_size_start = len(rolling_id_dictionary)
    vip_match_count = 0
    candidate_count = 0
    new_matched_count = 0
    new_partial_count = 0
    
    # Isolate current season using respective season columns
    sswy_season = df_unmatched_sswy[df_unmatched_sswy['season'] == season].copy()
    sofifa_season = df_unmatched_sofifa[df_unmatched_sofifa['sofifa_season'] == season].copy()
    
    if sswy_season.empty or sofifa_season.empty:
        continue

    # --------------------------------------------------------------------------
    # STEP A: THE VIP DICTIONARY BYPASS
    # --------------------------------------------------------------------------
    sswy_season['mapped_sofifa'] = sswy_season['soccersolver_id'].map(rolling_id_dictionary)
    vip_matches = sswy_season.dropna(subset=['mapped_sofifa'])
    vip_matches = vip_matches[vip_matches['mapped_sofifa'].isin(sofifa_season['sofifa_player_id'])]
    
    if not vip_matches.empty:
        vip_to_append = pd.merge(
            vip_matches, 
            sofifa_season, 
            left_on='mapped_sofifa', 
            right_on='sofifa_player_id', 
            suffixes=('_sswy', '_sofifa')
        )
        vip_to_append['match_score'] = 100 
        vip_to_append['match_type'] = 'VIP_BYPASS_SECONDARY'
        
        df_master_matched = pd.concat([df_master_matched, vip_to_append], ignore_index=True)
        
        sswy_season = sswy_season[~sswy_season['soccersolver_id'].isin(vip_to_append['soccersolver_id'])]
        sofifa_season = sofifa_season[~sofifa_season['sofifa_player_id'].isin(vip_to_append['sofifa_player_id'])]
        
        vip_match_count = len(vip_to_append)

    leftovers_count = len(sofifa_season)

    # --------------------------------------------------------------------------
    # STEP B: THE GATED HEURISTIC EVALUATION
    # --------------------------------------------------------------------------
    # GATE 1: The Merge (Normalized Nationality + First Letter)
    merged_pool = pd.merge(
        sswy_season, 
        sofifa_season, 
        on=['normalized_nationality', 'name_first_letter'], 
        suffixes=('_sswy', '_sofifa')
    )
    
    if not merged_pool.empty:
        # GATE 2: Physical Tolerances (Height OR Foot fallback)
        cond_height = (merged_pool['height_cm'] - merged_pool['sofifa_height_cm']).abs() <= 3
        cond_foot = merged_pool['height_cm'].isna() & (merged_pool['preferred_foot'] == merged_pool['sofifa_preferred_foot'])
        
        merged_pool = merged_pool[cond_height | cond_foot].copy()
        candidate_count = len(merged_pool)
        
        if not merged_pool.empty:
            # GATE 3: String Evaluation (using normalized names)
            merged_pool['name_score'] = merged_pool.apply(
                lambda row: fuzz.token_set_ratio(str(row['normalized_name_sswy']), str(row['normalized_name_sofifa'])), 
                axis=1
            )
            
            matched_mask = merged_pool['name_score'] > 85
            partial_mask = (merged_pool['name_score'] >= 75) & (merged_pool['name_score'] <= 85)
            
            new_matched = merged_pool[matched_mask].copy()
            new_partial = merged_pool[partial_mask].copy()
            
            if not new_matched.empty:
                new_matched['match_type'] = 'SECONDARY_HEURISTIC_MATCH'
                df_master_matched = pd.concat([df_master_matched, new_matched], ignore_index=True)
                
                # Update rolling dictionary using exact ID columns
                new_dict_entries = dict(zip(new_matched['soccersolver_id'], new_matched['sofifa_player_id']))
                rolling_id_dictionary.update(new_dict_entries)
                new_matched_count = len(new_matched)
                
            if not new_partial.empty:
                new_partial['match_type'] = 'SECONDARY_HEURISTIC_PARTIAL'
                df_master_partial = pd.concat([df_master_partial, new_partial], ignore_index=True)
                new_partial_count = len(new_partial)

    # --------------------------------------------------------------------------
    # LIVE TRACKER CONSOLE OUTPUT
    # --------------------------------------------------------------------------
    print(f"\n[{idx}/{total_seasons}] LIVE TRACKER: Season {season}")
    print("-" * 80)
    print(f" -> VIP Bypass:             {vip_match_count:<5} Matched (Current Dictionary Size: {dict_size_start})")
    print(f"    [Rookies/Leftovers pushed to Heuristics: {leftovers_count} Sofifa Players]")
    print(f" -> Gates 1 & 2 (Physical): {candidate_count:<5} Survived to String Evaluation")
    print(f" -> Gate 3 (String Score):  {new_matched_count:<5} Matched (>85%) | {new_partial_count} Partial (75-85%)")
    if new_matched_count > 0:
        print(f" [+] Added {new_matched_count} newly discovered pairs to the Rolling Dictionary.")

# ==============================================================================
# 3. THE RETROACTIVE LOOP-BACK (TIME MACHINE)
# ==============================================================================
print("\n" + "=" * 80)
print("Executing Retroactive Time Machine Pass...")

df_unmatched_sswy['retro_mapped_sofifa'] = df_unmatched_sswy['soccersolver_id'].map(rolling_id_dictionary)
retro_rescued_sswy = df_unmatched_sswy.dropna(subset=['retro_mapped_sofifa'])

if not retro_rescued_sswy.empty:
    retro_rescued_full = pd.merge(
        retro_rescued_sswy,
        df_unmatched_sofifa,
        left_on=['retro_mapped_sofifa', 'season'],
        right_on=['sofifa_player_id', 'sofifa_season'], # Aligning disparate season columns
        suffixes=('_sswy', '_sofifa')
    )
    
    retro_rescued_full['match_score'] = 100
    retro_rescued_full['match_type'] = 'RETROACTIVE_RESCUE_SECONDARY'
    df_master_matched = pd.concat([df_master_matched, retro_rescued_full], ignore_index=True)
    
    print(f" [*] Retroactive Sweep Complete: Rescued {len(retro_rescued_full)} historical records.")

# ==============================================================================
# 4. THE FINAL PURGE (Hygiene Maintenance)
# ==============================================================================
print("Purging Matched/Partial records from Unmatched datasets...")

resolved_sswy_ids = set(df_master_matched['soccersolver_id']).union(set(df_master_partial['soccersolver_id']))
resolved_sofifa_ids = set(df_master_matched['sofifa_player_id']).union(set(df_master_partial['sofifa_player_id']))

df_unmatched_sswy = df_unmatched_sswy[~df_unmatched_sswy['soccersolver_id'].isin(resolved_sswy_ids)]
df_unmatched_sofifa = df_unmatched_sofifa[~df_unmatched_sofifa['sofifa_player_id'].isin(resolved_sofifa_ids)]

df_unmatched_sswy.drop(columns=['retro_mapped_sofifa'], errors='ignore', inplace=True)

print("Secondary Pipeline Execution Complete.\n" + "=" * 80)
```

```python
import pandas as pd
import numpy as np
from thefuzz import fuzz

# ==============================================================================
# 1. LOAD DATA & SANITIZE DICTIONARY
# ==============================================================================
print("Loading datasets...")
df_partial = pd.read_csv("newdata/players/partial match/FINAL_PARTIAL_REVIEW.csv")
df_matched = pd.read_csv("newdata/players/matched/FINAL_MATCHED_MASTER.csv")
df_unmatched_sswy = pd.read_csv("newdata/players/no match/UNMATCHED_SSWY.csv")
df_unmatched_sofifa = pd.read_csv("newdata/players/no match/UNMATCHED_SOFIFA.csv")

# Ensure dictionary is loaded and sanitized to a flat 1D string mapping
dict_df = pd.read_csv("newdata/players/matched/FINAL_ID_DICTIONARY.csv")
rolling_id_dictionary = dict(zip(dict_df['soccersolver_id'].astype(str), dict_df['sofifa_player_id'].astype(str)))

# Ensure IDs are strings in the partial file for clean lookups
df_partial['soccersolver_id'] = df_partial['soccersolver_id'].astype(str).str.replace(r'\.0$', '', regex=True)
df_partial['sofifa_player_id'] = df_partial['sofifa_player_id'].astype(str).str.replace(r'\.0$', '', regex=True)

# ==============================================================================
# 2. EXTRACT MASTER TEAM DICTIONARY (Ground Truth)
# ==============================================================================
valid_teams = df_matched.dropna(subset=['team', 'sofifa_club_name'])
master_team_set = set(zip(valid_teams['team'].str.lower(), valid_teams['sofifa_club_name'].str.lower()))
print(f"Extracted {len(master_team_set)} verified Team-to-Club pairings from Master.")

# ==============================================================================
# 3. DEFINE THE FALLBACK EVALUATOR
# ==============================================================================
def evaluate_partial_row(row):
    sswy_name = str(row.get('normalized_name_sswy', row.get('name', ''))).lower()
    sof_short = str(row.get('normalized_name_sof', row.get('sofifa_name', ''))).lower()
    sof_full = str(row.get('sofifa_full_name', '')).lower()
    
    sswy_team = str(row.get('team', '')).lower()
    sof_team = str(row.get('sofifa_club_name', '')).lower()
    
    dob_sswy = str(row.get('clean_dob_sswy', 'nan'))
    dob_sof = str(row.get('sofifa_dob', 'nan'))
    
    nat_sswy = str(row.get('normalized_nationality_sswy', 'nan')).lower()
    nat_sof = str(row.get('normalized_nationality_sof', 'nan')).lower()
    
    foot_sswy = str(row.get('preferred_foot', 'nan')).lower()
    foot_sof = str(row.get('sofifa_preferred_foot', 'nan')).lower()
    
    h_sswy = row.get('height_cm')
    h_sof = row.get('sofifa_height_cm')
    height_match = False
    if pd.notna(h_sswy) and pd.notna(h_sof):
        height_match = abs(float(h_sswy) - float(h_sof)) <= 2

    # Calculate best name score
    best_name_score = max(fuzz.token_set_ratio(sswy_name, sof_short), fuzz.token_set_ratio(sswy_name, sof_full))
    
    # GLOBAL STEP 1: The Master Team Anchor
    if sswy_team != 'nan' and sof_team != 'nan' and (sswy_team, sof_team) in master_team_set:
        if best_name_score >= 70:
            return 'TEAM_ANCHOR'
            
    # GLOBAL STEP 2: The Fallback Logic
    has_dob = (dob_sswy != 'nan' and dob_sof != 'nan' and dob_sswy != '' and dob_sof != '')
    foot_match = (foot_sswy != 'nan' and foot_sswy == foot_sof)
    nat_match = (nat_sswy != 'nan' and nat_sswy == nat_sof)

    if has_dob:
        # CASE 1: DOB Present
        if dob_sswy == dob_sof and nat_match:
            if foot_match or height_match:
                if best_name_score >= 70:  # Name safety net
                    return 'CASE_1_DOB'
    else:
        # CASE 2: DOB Missing
        if nat_match and foot_match and height_match:
            if best_name_score >= 80:
                return 'CASE_2_NO_DOB'
                
    return 'REJECTED'

# ==============================================================================
# 4. CHRONOLOGICAL LOOP & EVALUATION (With Live Tracker)
# ==============================================================================
df_partial['rescue_status'] = 'PENDING'
seasons = sorted(df_partial['season'].dropna().unique())

print("\n" + "="*50)
print("STARTING SNOWBALL RESCUE ENGINE RUN")
print("="*50)

for idx, season in enumerate(seasons, 1):
    season_mask = df_partial['season'] == season
    total_season_records = season_mask.sum()
    
    # Global Step 0: VIP Bypass
    mapped_sofifa = df_partial.loc[season_mask, 'soccersolver_id'].map(rolling_id_dictionary)
    vip_mask = mapped_sofifa == df_partial.loc[season_mask, 'sofifa_player_id']
    num_vip = vip_mask.sum()
    
    # Apply VIP statuses
    df_partial.loc[season_mask & vip_mask, 'rescue_status'] = 'VIP_BYPASS'
    
    # Live tracker console header
    print(f"\nProcessing Season [{idx}/{len(seasons)}]: {season}")
    print(f" -> Total records in season: {total_season_records}")
    print(f" -> VIP Bypass matches found: {num_vip}")
    
    # Evaluate Leftovers
    leftover_mask = season_mask & ~vip_mask
    num_leftovers = leftover_mask.sum()
    
    if num_leftovers > 0:
        print(f" -> Running fallback heuristics on remaining {num_leftovers} rows...")
        df_partial.loc[leftover_mask, 'rescue_status'] = df_partial[leftover_mask].apply(evaluate_partial_row, axis=1)
        
        # Count outcomes for this season's fallback execution
        season_results = df_partial[leftover_mask]['rescue_status'].value_counts()
        for status, count in season_results.items():
            print(f"    - {status}: {count}")
            
        # Snowball Update: Add new approvals from this season to the dictionary
        new_approvals = df_partial[leftover_mask & (df_partial['rescue_status'] != 'REJECTED')]
        new_dict_entries = dict(zip(new_approvals['soccersolver_id'], new_approvals['sofifa_player_id']))
        rolling_id_dictionary.update(new_dict_entries)
    else:
        print(" -> No remaining rows to process via fallback heuristics.")

print("\n" + "="*50)
print("RUNNING RETROACTIVE SWEEP (TIME MACHINE)")
print("="*50)

# Phase 4: Retroactive Sweep (Time Machine)
retro_mask = df_partial['rescue_status'] == 'REJECTED'
num_initial_rejected = retro_mask.sum()
print(f"Initial rejected rows across all seasons: {num_initial_rejected}")

mapped_retro = df_partial.loc[retro_mask, 'soccersolver_id'].map(rolling_id_dictionary)
retro_success_mask = mapped_retro == df_partial.loc[retro_mask, 'sofifa_player_id']
num_retro_rescued = retro_success_mask.sum()

df_partial.loc[retro_mask & retro_success_mask, 'rescue_status'] = 'RETROACTIVE_RESCUE'
print(f" -> Successfully rescued via Retroactive Matured Dictionary: {num_retro_rescued}")

# ==============================================================================
# 5. FINAL ROUTING & SAVING (Collision-Proof)
# ==============================================================================
approved_mask = df_partial['rescue_status'] != 'REJECTED'
approved_records = df_partial[approved_mask].copy()
rejected_records = df_partial[~approved_mask].copy()

print("\n" + "="*50)
print("FINAL PIPELINE EXECUTION SUMMARY")
print("="*50)
print(approved_records['rescue_status'].value_counts())
print(f"Total True False-Positives (Sent to Unmatched): {len(rejected_records)}")
print("="*50)

# ROUTE 1: Append Approved to Master Matched
approved_records['match_type'] = 'PARTIAL_RESCUE_' + approved_records['rescue_status']
approved_records.drop(columns=['rescue_status'], inplace=True)
df_matched = pd.concat([df_matched, approved_records], ignore_index=True)

# ROUTE 2: Un-merge Rejected back to Unmatched Files (Collision-Proof mapping)
rejected_sswy = pd.DataFrame()
for col in df_unmatched_sswy.columns:
    if f"{col}_sswy" in rejected_records.columns:
        rejected_sswy[col] = rejected_records[f"{col}_sswy"]
    elif col in rejected_records.columns:
        rejected_sswy[col] = rejected_records[col]
    else:
        rejected_sswy[col] = np.nan

rejected_sofifa = pd.DataFrame()
for col in df_unmatched_sofifa.columns:
    if f"{col}_sof" in rejected_records.columns:
        rejected_sofifa[col] = rejected_records[f"{col}_sof"]
    elif f"{col}_sofifa" in rejected_records.columns:
        rejected_sofifa[col] = rejected_records[f"{col}_sofifa"]
    elif col in rejected_records.columns:
        rejected_sofifa[col] = rejected_records[col]
    else:
        rejected_sofifa[col] = np.nan

# Append back to respective unmatched files
df_unmatched_sswy = pd.concat([df_unmatched_sswy, rejected_sswy], ignore_index=True)
df_unmatched_sofifa = pd.concat([df_unmatched_sofifa, rejected_sofifa], ignore_index=True)

# Save all files
print("\nSaving updated Master and Unmatched files...")
df_matched.to_csv("newdata/players/matched/FINAL_MATCHED_MASTER.csv", index=False)
df_unmatched_sswy.to_csv("newdata/players/no match/UNMATCHED_SSWY.csv", index=False)
df_unmatched_sofifa.to_csv("newdata/players/no match/UNMATCHED_SOFIFA.csv", index=False)

# Empty out the Partial file and overwrite it
pd.DataFrame(columns=df_partial.columns).to_csv("newdata/players/partial match/FINAL_PARTIAL_REVIEW.csv", index=False)

# Save the updated dictionary
pd.DataFrame(list(rolling_id_dictionary.items()), columns=['soccersolver_id', 'sofifa_player_id']).to_csv("newdata/players/matched/FINAL_ID_DICTIONARY.csv", index=False)

print("Process Complete. The Partial File has been cleanly drained.")
```
