# Merging Report

## PLAYERS

Notation:
- UPPERCASE = MATCHED ACROSS THAT DATASET
- lowercase = unmatched across that dataset

```python
Player_SS_WY_SOFIFA=pd.read_csv('newdata/players/matched/FINAL_MATCHED_MASTER.csv') #matched across 3 sources
Player_SS_WY_sofifa=pd.read_csv('newdata/players/no match/UNMATCHED_SSWY.csv')      #matched across SS and WY but not sofifa
Player_ss_wy_SOFIFA=pd.read_csv('newdata/players/no match/UNMATCHED_SOFIFA.csv')    #sofifa unmatched players with SS and WY
Player_ss=pd.read_csv('data/unified_tables/players/no match/soccersolver_unmatched.csv')    #soccercolver players that did not match with Wyscout players
Player_wy=pd.read_csv('data/unified_tables/players/no match/wyscout_unmatched.csv')     #wyscout players that did not match with ss
```

```python
print('='*50)
print('DIAGNOSTIC REPORT FOR PLAYERS')
print('='*50)
print("SOURCES: Soccersolver Data, Wyscout Data, Sofifa data")
print("DISCRIPTION: SEASONAL RECORDS FROM 2019/20 TO 2025/26")
print('='*50)
print(f'Number of player records matched across all 3 sources: {len(Player_SS_WY_SOFIFA)}')
print(f'Number of player records matched across Soccersolver and Wyscout but not sofifa: {len(Player_SS_WY_sofifa)}')
print(f'Number of Player records from Sofifa unmatched with SoccerSolver and Wyscout: {len(Player_ss_wy_SOFIFA)}')
print(f'Number of unmatched soccersolver player records: {len(Player_ss)}')
print(f'Number of unmatched Wyscout records {len(Player_wy)}')
print(f'Total records processed: {(len(Player_SS_WY_SOFIFA))+(len(Player_SS_WY_sofifa))+(len(Player_ss_wy_SOFIFA))+(len(Player_ss))+(len(Player_wy))}')
print('='*50)
print(' END OF PLAYER REPORT')
```
